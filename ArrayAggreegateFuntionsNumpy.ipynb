# Array Aggregate functions in Numpy
Aggregate functions are the helper function in NumPy that help to group the values of the array together to give the single summary value.
Some of the aggregate function in Python are:
- sum(): Sum of all the elements in an array
- min(): Find the smallest element in an array
- max(): Find the largest element in an array
- mean(): The mean value of the array
- median(): The median value of the array
- corrcoef(): To find the correlation coefficient
- std(): Finding the standard deviation.


```python
import numpy as np

a = [1,2,3,4]
#To find the aggregate sum of the array
sum = np.sum(a)
print("Sum of array elements is",sum)

#To find the minimum value in an array
min = np.min(a)
print("Minimum of array elements is",min)

#To find the maximum value in the array
max = np.max(a)
print("Maximum of array elements is",max)

#To find the mean value of the array
mean = np.mean(a)
print("Mean of array elements is",mean)

#To find the median value of the array.
med = np.median(a)
print("Median of array elements is",med)

#To find the correaltion coefficiant 
cor = np.corrcoef(a)
print("Correaltion coefficiant of array elements is",cor)

#To find the standard deviation 
std = np.std(a)
print("standard deviation of array elements is",std)
```
