# # Map

Being a higher-order function, the map function takes another function and an iterable (e.g., a list, set, tuple) as input, applies the function to the iterable, and returns an output. It’s syntax is defined as follows:

    map(function, iterable)
    
 map(cook, [🐷,🌽,🥚 ,🍠]) → [🥓,🍿,🍳,🍟]

```python
def add4(x):
    return x+4
list1 = [4,6,7,8,9]

list2 = list(map(add4,list1))
list2
```

```python
list3 = list(map(lambda x:x+4,list1))
list3
```

```python
set_of_strings = ['abc','def','xyz']
string_map_22 = list(map(lambda my_string: my_string + '_2022', set_of_strings))
string_map_22
```

# Filter()

Similar to map(), the filter() higher-order function takes a function and an iterable as inputs. The function in case needs to be of a Boolean nature, returning True/False values corresponding to the filter conditions. As output, it returns a subset of the input data that meets the conditions stipulated by the function.

filter(non_vegan, [🥓,🍿,🍳,🍟]) → [🥓,🍳]

```python
def oddeven(x):
    if x%2 == 0:
        return True
    else:
        return False
    
list1 = [4,5,6,7,8,9]
evenlist = list(filter(oddeven,list1))
evenlist
```

```python
list1 = [4,5,6,7,8,9]
evenlist = list(filter(lambda x:True if x%2==0 else False,list1))
evenlist
```

# Reduce
It returns a single value (i.e., it reduces the input to a single element). Commonly, this would be something like the sum of all elements in a list.
Example
reduce(mix, [🥬,🥕,🍅,🥒]) →🥗

```python
from functools import reduce
def sum(x,y):
    return x+y

list1 = [6,7,8,9]
s = reduce(sum,list1)
s
```
