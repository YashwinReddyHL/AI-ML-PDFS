# Create a dataframe with following data.

```python
import numpy as np 
#importing pandas
import pandas as pd
```

```python
employee_details = {'FirstName': ['Aryan','Rohan','Riya','Yash','Siddhant','xyz','pqr'], 
        'LastName': ['Singh','Agarwal','Shah','Bhatia','Khanna','abc','mno'],  
        'Type': ['Full-time Employee','Intern','Full-time Employee','Part-time Employee',
                 'Full-time Employee','Intern','Full-time Employee'],  
        'Department': ['Administration','Technical','Administration','Technical',
                       'Administration','Administration','Technical'],
        'YoE':[2,3,5,7,6,1,7],
        'Salary':[20000,5000,10000,10000,20000,4000,30000]}
employee_df = pd.DataFrame(employee_details)
employee_df
```

# Make a pivot table which shows average salary of each type of employee for each department.

```python
#multiple indexes
table = pd.pivot_table(employee_df,index=['Type','Department'],values='Salary')
table
```

# Make a pivot table which shows the sum and mean of the salaries of each type of employee and the number of employees of each type.

```python
pivot = pd.pivot_table(
    data=employee_df,
    index='Type',
    values='Salary',
    aggfunc=['mean', 'sum']
)

print(pivot)
```
